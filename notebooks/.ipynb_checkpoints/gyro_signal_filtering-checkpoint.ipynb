```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import butter, lfilter
import biosppy
from scipy.fftpack import fft

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import biosppy
from biosppy.signals import ppg
from biosppy.signals.tools import filter_signal
import plotly.graph_objs as go
from scipy.signal import find_peaks, butter, filtfilt
import librosa
```

```python
df = pd.read_csv('data/train/california-00011978-right-sync.csv')
#df = pd.read_csv('data/adhd_train/adhd-KKI_004-left-sync.csv')

df['sleep_label'] = df['sleep_stage'].apply(lambda x: 0 if x == 'WK' else 1)  # 0: awake, 1: sleep
df = df.dropna(subset=['gyroscopeX', 'gyroscopeY', 'gyroscopeZ', 'sleep_label'])
df['formatted_time'] = pd.to_datetime(df['unixTimes'], unit='ms')
```

```python
df
```

```python
# Plot with Plotly
fig = go.Figure()

# Add the filtered gyroscope data traces
fig.add_trace(go.Scatter(x=df['unixTimes'], y=df['gyroscopeX'], mode='lines', name='Gyroscope X'))
fig.add_trace(go.Scatter(x=df['unixTimes'], y=df['gyroscopeY'], mode='lines', name='Gyroscope Y'))
fig.add_trace(go.Scatter(x=df['unixTimes'], y=df['gyroscopeZ'], mode='lines', name='Gyroscope Z'))

# Update layout for better visualization
fig.update_layout(
    title='Filtered Gyroscope Data',
    xaxis_title='Time (unixTimes)',
    yaxis_title='Angular Velocity (°/s)',
    hovermode='x unified'
)

# Show the plot
fig.show()
```

```python
def butter_lowpass_filter(data, cutoff, fs, order=4):
    nyquist = 0.5 * fs
    normal_cutoff = cutoff / nyquist
    b, a = butter(order, normal_cutoff, btype='low', analog=False)
    y = filtfilt(b, a, data)
    return y

sampling_rate = 25  # Example sampling rate in Hz
cutoff_frequency = 5  # Hz
df['gyroX_filtered'] = butter_lowpass_filter(df['gyroscopeX'], cutoff_frequency, sampling_rate)
df['gyroY_filtered'] = butter_lowpass_filter(df['gyroscopeY'], cutoff_frequency, sampling_rate)
df['gyroZ_filtered'] = butter_lowpass_filter(df['gyroscopeZ'], cutoff_frequency, sampling_rate)
```

```python
# Plot with Plotly
fig = go.Figure()

# Add the filtered gyroscope data traces
fig.add_trace(go.Scatter(x=df['unixTimes'], y=df['gyroX_filtered'], mode='lines', name='Gyroscope X'))
fig.add_trace(go.Scatter(x=df['unixTimes'], y=df['gyroY_filtered'], mode='lines', name='Gyroscope Y'))
fig.add_trace(go.Scatter(x=df['unixTimes'], y=df['gyroZ_filtered'], mode='lines', name='Gyroscope Z'))

# Update layout for better visualization
fig.update_layout(
    title='Filtered Gyroscope Data',
    xaxis_title='Time (unixTimes)',
    yaxis_title='Angular Velocity (°/s)',
    hovermode='x unified'
)

# Show the plot
fig.show()
```
