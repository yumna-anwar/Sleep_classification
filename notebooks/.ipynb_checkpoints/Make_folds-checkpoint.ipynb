```python
import os
import shutil
import re
import random
```

```python

# Function to get numeric ID from the filename
def get_numeric_id(filename):
    match = re.search(r'\d+', filename)
    return int(match.group()) if match else None

# Function to split files into 3 folds
def split_into_folds(files, num_folds=3):
    random.shuffle(files)
    
    # Group files by ID
    file_groups = {}
    for file in files:
        file_id = get_numeric_id(file)
        if file_id:
            if file_id not in file_groups:
                file_groups[file_id] = []
            file_groups[file_id].append(file)

    # Prepare the fold lists
    folds = [[] for _ in range(num_folds)]

    # Distribute files across folds
    for i, (file_id, group) in enumerate(file_groups.items()):
        folds[i % num_folds].extend(group)

    return folds

# Function to create folders and move files into them
def create_folds_and_move_files(folder_path, folds):
    # Create fold directories if they don't exist
    for i in range(len(folds)):
        fold_dir = os.path.join(folder_path, f'fold{i+1}')
        os.makedirs(fold_dir, exist_ok=True)

        # Move the files into the respective fold directory
        for file in folds[i]:
            src = os.path.join(folder_path, file)
            dst = os.path.join(fold_dir, file)
            shutil.move(src, dst)
            print(f'Moved {file} to {fold_dir}')

# Main function to process files
def split_and_move_files_into_folds(folder_path):
    # List all CSV files in the folder
    all_files = [f for f in os.listdir(folder_path) if f.endswith('.csv')]

    # Split files into 3 folds
    folds = split_into_folds(all_files, num_folds=5)

    # Create the fold directories and move the files
    create_folds_and_move_files(folder_path, folds)
```

```python
# Usage example
folder_path = '/home/yanwar/Desktop/Sleep/data/folds'
split_and_move_files_into_folds(folder_path)
```
